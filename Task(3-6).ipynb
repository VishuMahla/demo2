```python
import numpy as np
```

```python
# Create a NumPy array containing 10, 20, 30, 40, 50
arr1 = np.array([i*10 for i in range(1,6)])
print(arr1)
```

```python
#Exercise 2: Check Array Properties
arr = np.array([[1, 2, 3],[4, 5, 6]])
#Print Shape
print("Shape: ", arr.shape)
#Print Dimensions
print("Dimension: ", arr.ndim)
# Print data type
print("Data Type:", arr.dtype)
```

```python
#Exercise 3: Generate Zeros Array
# Create a 3x4 array filled with zeros
arr = np.zeros((3,4))
print(arr)
```

```python
# Create an array containing even numbers from 2 to 20
arr = np.array([i for i in range(2,21,2)])
print(arr)
#or
arr = np.array([i for i in range(2,21) if not i%2]) # or if i%2 == 0
print(arr)

#or 
arr = np.arange(2,21,2)
print(arr)
```

```python
# Exercise 5: Reshape an Array
# Convert the array into 3 rows and 4 columns
arr = np.arange(1, 13)
print(arr.reshape(3,4))
```

```python
#Exercise 6: Find Maximum and Minimum
import numpy as np
arr = np.array([25, 40, 10, 90, 65])
# Print maximum value
print("Max =", arr.max())
# Print minimum value
print("Min =", arr.min())
```

```python
#Exercise 7: Array Arithmetic
import numpy as np
arr1 = np.array([1, 2, 3, 4])
arr2 = np.array([10, 20, 30, 40])
# Add both arrays
result = arr1 + arr2
print(result)
```

```python
#Exercise 8: Filter Elements
import numpy as np
arr = np.array([5, 12, 18, 7, 25, 30])
# Select elements greater than 15
result = arr[arr > 15]
print(result)
```

```python
#Exercise 9: Calculate Mean
import numpy as np
arr = np.array([10, 20, 30, 40, 50])
# Calculate average value
avg = arr.mean()
print(avg)
```

```python
# Exercise 10: Matrix Multiplication
import numpy as np
A = np.array([[1, 2],
 [3, 4]])
B = np.array([[5, 6],
 [7, 8]])
# Perform matrix multiplication
C = A @ B 
print(C)
#orrr
C = A.dot(B)
print(C)
```
