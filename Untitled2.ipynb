```python
# 1. Function to print Fibonacci Series up to n terms
def fibonacci(n):
    a, b = 0, 1

    for i in range(n):
        print(a, end=" ")
        a, b = b, a + b
fibonacci(1)
```

```python

# 2. Function to check whether a string is palindrome or not
def string_palindrome(s):
    if s == s[::-1]:
        print("Palindrome")
    else:
        print("Not Palindrome")
string_palindrome("hello")
```

```python
# 3. Function to check whether a number is palindrome or not
def number_palindrome(num):
    original = num
    reverse = 0

    while num > 0:
        digit = num % 10
        reverse = reverse * 10 + digit
        num = num // 10

    if original == reverse:
        print("Palindrome Number")
    else:
        print("Not Palindrome Number")
number_palindrome(121)
```

```python
# 4. Function to print sum of digits of a number
def sum_of_digits(num):
    total = 0

    while num > 0:
        digit = num % 10
        total += digit
        num = num // 10

    print("Sum of digits:", total)
sum_of_digits(1234)
```

```python
# 5. Function to print count of digits in a number
def count_digits(num):
    count = 0

    while num > 0:
        num = num // 10
        count += 1

    print("Count of digits:", count)


# Example
count_digits(12345)
```

```python
# 6. Function to check Armstrong number
def armstrong(num):
    original = num
    power = len(str(num))
    total = 0

    while num > 0:
        digit = num % 10
        total += digit ** power
        num = num // 10

    if original == total:
        print("Armstrong Number")
    else:
        print("Not Armstrong Number")
armstrong(153)
```

```python
# 7. Function to check Perfect number
def perfect_number(num):
    total = 0

    for i in range(1, num):
        if num % i == 0:
            total += i

    if total == num:
        print("Perfect Number")
    else:
        print("Not Perfect Number")
perfect_number(6)
```

```python
# 8. Function to return factorial of a number
def factorial(num):
    fact = 1

    for i in range(1, num + 1):
        fact *= i

    return fact


# Example
print(factorial(5))
```

```python
# 9. Function to check Strong number
def strong_number(num):
    original = num
    total = 0

    while num > 0:
        digit = num % 10

        fact = 1
        for i in range(1, digit + 1):
            fact *= i

        total += fact
        num = num // 10

    if original == total:
        print("Strong Number")
    else:
        print("Not Strong Number")
strong_number(145)
```

```python
# 10. Function to count vowels in a string
def count_vowels(s):
    count = 0

    for ch in s.lower():
        if ch in "aeiou":
            count += 1

    print("Number of vowels:", count)

count_vowels("india")
```
