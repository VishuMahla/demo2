```python
import numpy as np #gives n dimensional array
arr = np.array([1,2,3])
print(arr*2);
```

```python
#addition of 2 arrays
arr1 = np.array([1,2,3,4])
arr2 = np.array([3,8,2,9])
print(arr1 + arr2);
```

```python
arr3 = np.array([[1,2,4],[4,6,9],[8,4,1]])
print(arr3)

```

```python
arr4 = np.array(list([0 for i in range(10)] for i in range(10)))
print(arr4)
```

```python
arr5 = np.array(list([x + y for i in range(10)] for x in range(5, 1, -1) for y in range (4)))
print(arr5)
```

```python
#useful array creation methods
a = np.zeros(3)
print(a)
type(a)
b = np.ones(3)
print(b)
c = np.arange(10)
print(type(c))
print(c)
d = np.linspace(0,1,6)
print(d)
e = np.eye(3)
print(e)
print("Total Bytes: ", e.nbytes)
```

```python
#properties 
shape return dimestions fo the array(rows, columsn)
size total no of elements
ndim return no of dimension of array
dtype return data type of array elements
itemsize return size of each element in bytes
nbytes return total memory used by the array in bytes
T transpose of array
real return real part
imag return imag part
flat return iterator over all array elements
```
